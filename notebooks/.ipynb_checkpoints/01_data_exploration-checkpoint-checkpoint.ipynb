```python
# Resume quality metrics
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
fig.suptitle('Resume Quality Indicators', fontsize=13, fontweight='bold')

# Has summary
summary_counts = df_resumes['has_summary'].value_counts()
axes[0].bar(['Has Summary', 'No Summary'], summary_counts.values,
             color=['#22c55e','#ef4444'], alpha=0.85, edgecolor='white')
axes[0].set_title('Professional Summary Presence')
axes[0].set_ylabel('Count')
for i, v in enumerate(summary_counts.values):
    axes[0].text(i, v+0.5, str(v), ha='center', fontweight='bold')

# Quantified achievements
qa_counts = df_resumes['has_quantified_achievements'].value_counts()
axes[1].bar(['Has Metrics', 'No Metrics'], qa_counts.values,
             color=['#6c63ff','#f59e0b'], alpha=0.85, edgecolor='white')
axes[1].set_title('Quantified Achievements in Resumes')
axes[1].set_ylabel('Count')
for i, v in enumerate(qa_counts.values):
    axes[1].text(i, v+0.5, str(v), ha='center', fontweight='bold')

plt.tight_layout()
plt.show()
print(f"\nOnly {qa_counts[True]/len(df_resumes)*100:.1f}% of resumes have quantified achievements!")
print(f"This is a major gap our AI rewriter addresses.")
```
